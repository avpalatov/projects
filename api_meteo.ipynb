```python
import pandas as pd
import requests
import datetime
import pyarrow

# Передаем сегодняшнюю дату 
load_date = datetime.date.today()
load_date_start = "2025-11-15"
load_date_end = datetime.date.today() - datetime.timedelta(days=1)


# отправляем запрос на сервер и получаем ответ в response
response = requests.get(
    "https://api.open-meteo.com/v1/forecast",
    params={
        "latitude": 55.0755,
        "longitude": 47.1506,
        "hourly": ["temperature_2m", "rain"],
        "start_date": load_date_start,
        "end_date": load_date_end,
    },
)
# Проверяем ошибки
response.raise_for_status()
# Передаем ответ от сервера в формате json
order_data_dict = response.json()
# Выводим результат
# print(order_data_dict)

df = pd.DataFrame({
    "datetime": order_data_dict["hourly"]["time"],
    "temperatura": order_data_dict["hourly"]["temperature_2m"],
    "rain": order_data_dict["hourly"]["rain"],
    "update_at": load_date
})

# df["fetched_at"] = load_date

# print(df)

# Генерируем имя файла
filename_csv = f"avpalatov_api_meteo_response_{load_date}.csv"
filename_parquet = f"avpalatov_api_meteo_response_{load_date}.parquet"

# Преобразуем в Датафрейм
# df = pd.json_normalize(order_data_dict)

# Сохранить в CSV
if len(df) > 0:
    # Сохранение в файл CSV без индекса
    df.to_csv(filename_csv, sep=';', header=True, index=False) #csv
    df.to_parquet(filename_parquet, index=False) #parquet
```
